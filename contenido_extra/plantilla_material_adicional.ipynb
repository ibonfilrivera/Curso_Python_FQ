## Densidad de líquidos

La densidad relaciona la masa de una sustancia con el volumen que ocupa: $\rho = m / V$.

```python
# Una celda de código normal: se copia tal cual en el notebook del estudiante
masa_agua = 49.85      # g
volumen_agua = 50.0    # mL
print(f"Densidad del agua: {masa_agua / volumen_agua:.3f} g/mL")
```

### Ejercicio adicional: densidad del etanol

Una probeta con 50.0 mL de etanol tiene una masa de 39.45 g. Calcula la densidad del etanol y
guárdala en `densidad`.

```python
#@ejercicio Densidad del etanol
#@revisar densidad g/mL
#@pista La densidad es la masa dividida entre el volumen.
masa = 39.45       # g
volumen = 50.0     # mL
densidad = masa / volumen
print(f"Densidad del etanol: {densidad:.3f} g/mL")
```

### Ejercicio adicional: corrección por el blanco

Resta la absorbancia del blanco (0.012) a las tres lecturas y guarda el resultado en
